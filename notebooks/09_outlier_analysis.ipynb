```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("fraud_detection_dashboard_data.csv")
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(y=df["Amount"])

plt.title("Outlier Detection - Transaction Amount")
plt.ylabel("Amount")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(y=df["Amount_log"])

plt.title("Outlier Detection - Log Transformed Amount")
plt.ylabel("Log Amount")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="Class",
    y="Amount"
)

plt.title("Transaction Amount by Class")
plt.xlabel("Class")
plt.ylabel("Amount")

plt.show()
```

```python
Q1 = df["Amount"].quantile(0.25)
Q3 = df["Amount"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["Amount"] < lower) | (df["Amount"] > upper)]

print("Number of Outliers:", len(outliers))
print("Percentage:", round(len(outliers)/len(df)*100,2), "%")
```
